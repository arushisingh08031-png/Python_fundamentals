```python
num = int(input("Enter a number: "))

factorial = 1
temp = num

while temp > 0:
    factorial *= temp
    temp -= 1

print(f"The factorial of {num} is {factorial}")
```

```python
num = int(input("Enter a number: "))
count = 0

temp = abs(num)

if temp == 0:
    count = 1
else:
    while temp > 0:
        count += 1
        temp = temp // 10  

print(f"The number of digits in {num} is {count}")
```

```python
n = int(input("Enter the number of terms: "))
a, b = 0, 1
count = 0

if n <= 0:
    print("Please enter a positive integer.")
elif n == 1:
    print(f"Fibonacci sequence up to {n} term: {a}")
else:
    print("Fibonacci sequence:")
    while count < n:
        print(a, end=" ")
        nth = a + b
        
        a = b
        b = nth
        count += 1
    print()  
```
